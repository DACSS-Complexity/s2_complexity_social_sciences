## Sensibilidad a las condiciones iniciales — versión combinada

Las dos celdas originales calculaban **la misma trayectoria logística** dos veces
(una vez para superponer las dos trayectorias, otra vez para graficar su distancia),
cada una con su propio par de sliders `r` / `ε`.

Aquí se calcula la trayectoria **una sola vez** y se generan las dos vistas
(trayectorias superpuestas y distancia) a partir del mismo resultado, controladas
por un único par de sliders y apiladas verticalmente con `alt.vconcat`.

Se agregó además un botón (`ToggleButton`) que aplica el logaritmo **natural**
(`ln`) a los datos del panel de distancia (no solo un cambio de escala del eje).
Esto es intencional: la fórmula de referencia es

$$\ln|x'_t - x_t| \approx \lambda t + c$$

así que, con el toggle activo, la **pendiente** de la curva resultante es una
estimación directa del exponente de Lyapunov `λ`. Con `r` en zona estable
(por ejemplo `r=2.51`) la pendiente sale negativa (las trayectorias convergen);
con `r` en zona caótica (por ejemplo `r=3.9`) sale positiva (las trayectorias
se separan).

Cuando las dos trayectorias colapsan al mismo valor exacto en `float64`
(algo que ocurre en régimen convergente, por límite de precisión numérica),
`ln(0)` no existe: en vez de inventar un piso arbitrario o cortar la curva
ahí, esos puntos puntuales se **saltean** y la curva sigue dibujándose con
los puntos válidos que vengan después. Esos huecos son un límite de
precisión de la máquina, no un cambio real en la dinámica.

**Sobre el número de `λ`:** ajustar una pendiente directamente sobre la
diferencia entre dos trayectorias resultó ser poco confiable en general
(por ejemplo, en regímenes con atractor periódico -período 2, período 4,
etc.- el "transiente" a descartar depende del período y no hay un número
fijo de pasos que funcione para todo `r`). Por eso el `λ` que se muestra
en el título del panel superior se calcula por un método **independiente
y robusto**: el promedio de `ln|f'(xₜ)|` sobre **una sola** trayectoria
larga (no dos), descartando un transiente fijo generoso. Al no restar dos
números casi iguales, este método no sufre el problema de precisión y
funciona igual de bien para puntos fijos, ciclos periódicos y regímenes
caóticos, sin ajustes por caso. El gráfico de distancia entre las dos
trayectorias (con los puntos coloreados) queda como **apoyo visual e
intuitivo únicamente** -- no se usa para calcular ningún número.

```python
import numpy as np
import pandas as pd
import altair as alt
import ipywidgets as widgets

from IPython.display import display, clear_output


# -----------------------------------
# Logistic trajectory
# -----------------------------------

def logistic_trajectory(x0, r, n):
    values = [x0]
    for _ in range(n):
        values.append(r * values[-1] * (1 - values[-1]))
    return values


def lambda_single_trajectory(r, n_transient=500, n_avg=2000, x0=0.4):
    """
    Robust numerical estimate of the Lyapunov exponent, following the
    analytic definition directly:

        lambda(r) = lim_{n->inf} (1/n) * sum_{t=1}^{n} ln|f'(x_t)|

    This works on a SINGLE trajectory (not a pair of nearby ones), so it
    never has to subtract two nearly-equal floats -- there is no precision
    floor to hit, no attractor period to detect, and no transient length
    to guess per-regime. A fixed, generous transient (n_transient) is safe
    here precisely because we're not differencing two close numbers.
    """
    x = x0
    for _ in range(n_transient):
        x = r * x * (1 - x)

    total = 0.0
    for _ in range(n_avg):
        deriv = abs(r * (1 - 2 * x))
        total += np.log(deriv) if deriv > 0 else -50.0  # deriv=0 only at x=0.5 exactly, measure zero
        x = r * x * (1 - x)

    return total / n_avg


# Fixed number of initial iterations treated as "transient" and excluded
# from the lambda estimate. Not user-adjustable on purpose -- the goal here
# is just to make the transient/asymptotic distinction visible, not to let
# students tune it (that would need its own careful discussion).
N_TRANSIENT_DISCARD = 5


# -----------------------------------
# Shared controls
# -----------------------------------

r_slider = widgets.FloatSlider(
    value=3.7,
    min=2.5,
    max=3.99,
    step=0.01,
    description="r:",
    continuous_update=False,
    readout_format=".2f"
)

epsilon_slider = widgets.FloatLogSlider(
    value=0.001,
    base=10,
    min=-5,      # 10^-5
    max=-1,      # 10^-1
    step=0.1,
    description="\u03b5:",
    continuous_update=False
)

log_scale_toggle = widgets.ToggleButton(
    value=False,
    description="Log scale (distance)",
    tooltip="Toggle log scale on the distance panel"
)

plot_output = widgets.Output()


# -----------------------------------
# Draw both charts (stacked) from one shared computation
# -----------------------------------

def draw_charts(*args):

    r = r_slider.value
    epsilon = epsilon_slider.value

    x0_1 = 0.10
    x0_2 = x0_1 + epsilon
    n_iter = 50

    traj_1 = np.array(logistic_trajectory(x0_1, r, n_iter))
    traj_2 = np.array(logistic_trajectory(x0_2, r, n_iter))
    difference = np.abs(traj_1 - traj_2)

    t = np.arange(n_iter + 1)

    # --- Data for the trajectories panel ---
    df_traj = pd.DataFrame({
        "t": np.concatenate([t, t]),
        "x": np.concatenate([traj_1, traj_2]),
        "Initial condition": (
            [f"x\u2080 = {x0_1:.6f}"] * (n_iter + 1)
            + [f"x\u2080 = {x0_2:.6f}"] * (n_iter + 1)
        )
    })

    # --- Data for the difference panel ---
    use_log = log_scale_toggle.value

    # Robust lambda estimate: computed independently of this plot (single
    # trajectory, generous fixed transient -- see function docstring for
    # why this is safe here but a fixed discard on the *two-trajectory*
    # difference below is not). This works the same way for a fixed point,
    # a period-p cycle, or a chaotic attractor, with no per-regime tuning.
    lambda_robust = lambda_single_trajectory(r)

    if use_log:
        # ln(0) doesn't exist. Instead of flooring with an arbitrary small
        # number (fake plateau) or cutting the curve at the first zero
        # (which would hide any later non-zero values, e.g. in a chaotic
        # regime where distance can dip to float64 zero and later grow
        # again), we simply drop the exact-zero points and keep plotting
        # everything else -- gaps in the line mark precision limits, not
        # a change in the dynamics.
        valid = difference != 0.0
        t_used = t[valid]
        diff_used = difference[valid]
        plotted_difference = np.log(diff_used)

        # Purely visual marker of "still approaching the attractor" vs.
        # "already on it" -- NOT used to fit anything anymore. Fitting a
        # slope directly on this two-trajectory difference is unreliable
        # in general (see lambda_robust above for why), so this coloring
        # is only meant to build intuition about transient vs. asymptotic
        # behavior, not to produce a number.
        point_role = np.where(
            t_used < N_TRANSIENT_DISCARD,
            "Transient (still approaching the attractor)",
            "Asymptotic (on the attractor)"
        )

        df_diff = pd.DataFrame({
            "t": t_used,
            "difference": plotted_difference,
            "raw_difference": diff_used,
            "role": point_role
        })
        n_skipped = int((~valid).sum())
    else:
        df_diff = pd.DataFrame({
            "t": t,
            "difference": difference,
            "raw_difference": difference,
            "role": "n/a"
        })
        n_skipped = 0

    # --- Top panel: the two trajectories ---
    chart_traj = (
        alt.Chart(df_traj)
        .mark_line(point=True, strokeWidth=2)
        .encode(
            x=alt.X("t:Q", title="Iteration"),
            y=alt.Y("x:Q", title="x\u209c", scale=alt.Scale(domain=[0, 1])),
            color=alt.Color("Initial condition:N", title="Initial condition"),
            strokeDash=alt.StrokeDash("Initial condition:N", legend=None),
            tooltip=[
                "Initial condition:N",
                alt.Tooltip("t:Q", title="Iteration"),
                alt.Tooltip("x:Q", format=".8f")
            ]
        )
        .properties(
            width=750,
            height=280,
            title=(
                f"Trajectories \u2014 r = {r:.2f}, \u03b5 = {epsilon:.6f}  "
                f"\u2014  robust \u03bb(r) \u2248 {lambda_robust:.4f}"
            )
        )
    )

    # --- Bottom panel: their absolute distance ---
    diff_title = (
        "ln(|x\u209c\u00b9 \u2212 x\u209c\u00b2|)"
        if use_log else
        "|x\u209c\u00b9 \u2212 x\u209c\u00b2|"
    )

    diff_chart_title = ["Distance between trajectories (qualitative only \u2014 see lambda above)"]
    if n_skipped > 0:
        diff_chart_title.append(f"{n_skipped} point(s) at exact 0 skipped (precision limit, not dynamics)")
    if use_log:
        diff_chart_title.append(
            f"Gray = still approaching the attractor (first {N_TRANSIENT_DISCARD} steps, purely illustrative)"
        )


    if use_log:
        color_enc = alt.Color(
            "role:N",
            title="Point role",
            scale=alt.Scale(
                domain=["Asymptotic (on the attractor)", "Transient (still approaching the attractor)"],
                range=["firebrick", "lightgray"]
            )
        )
    else:
        color_enc = alt.value("firebrick")

    # The connecting line stays a single continuous trace (uncolored by
    # category) so the curve reads as one line; only the points on top are
    # colored by role. Coloring the line mark itself would split it into
    # two disconnected series, one per color.
    line_color = "darkgray" if use_log else "firebrick"

    diff_line = (
        alt.Chart(df_diff)
        .mark_line(strokeWidth=2, color=line_color)
        .encode(
            x=alt.X("t:Q", title="Iteration"),
            y=alt.Y("difference:Q", title=diff_title),
            order="t:Q"
        )
    )

    diff_points = (
        alt.Chart(df_diff)
        .mark_point(filled=True, size=50)
        .encode(
            x=alt.X("t:Q", title="Iteration"),
            y=alt.Y("difference:Q", title=diff_title),
            color=color_enc,
            tooltip=[
                alt.Tooltip("t:Q", title="Iteration"),
                alt.Tooltip("raw_difference:Q", title="Distance", format=".8f"),
                alt.Tooltip("difference:Q", title="ln(Distance)" if use_log else "Distance", format=".6f")
            ]
        )
    )

    chart_diff = (
        (diff_line + diff_points)
        .properties(
            width=750,
            height=280,
            title=diff_chart_title
        )
    )

    combined = alt.vconcat(chart_traj, chart_diff).resolve_scale(color="independent")

    with plot_output:
        clear_output(wait=True)
        display(combined)


# -----------------------------------
# Connect sliders (shared by both plots)
# -----------------------------------

r_slider.observe(draw_charts, names="value")
epsilon_slider.observe(draw_charts, names="value")
log_scale_toggle.observe(draw_charts, names="value")

display(
    widgets.HBox([r_slider, epsilon_slider, log_scale_toggle]),
    plot_output
)

draw_charts()
```
