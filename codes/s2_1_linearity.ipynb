```python
import numpy as np
import pandas as pd
import altair as alt


# The rule: x(t+1) = r * x(t)
def linear_series(r, x0=0.10, T=25):

    x = np.empty(T + 1)
    x[0] = x0                  # initial state

    for t in range(T):
        x[t + 1] = r * x[t]   # apply the rule again

    return x


# ---- Simulation settings ----
x0 = 0.10                     # where we start
T = 25                        # how many iterations

r_min, r_max, r_step = 0.00, 1.50, 0.01

# Values of r that we want to explore
r_values = np.round(
    np.arange(r_min, r_max + 1e-12, r_step), 2
)

t = np.arange(T + 1)


# ---- Run the model for each value of r ----
rows = []

for r in r_values:

    xs = linear_series(r, x0=x0, T=T)

    for ti, xi in zip(t, xs):
        rows.append((r, ti, xi))


df = pd.DataFrame(rows, columns=["r", "t", "x"])


# ---- Interactive slider for r ----
r_param = alt.param(
    name="r",
    value=1.02,
    bind=alt.binding_range(
        min=r_min,
        max=r_max,
        step=r_step,
        name="r = "
    )
)


# ---- Plot the selected trajectory ----
base = (
    alt.Chart(df)
    .add_params(r_param)
    .transform_filter(alt.datum.r == r_param)
    .encode(
        x=alt.X("t:Q", title="t"),
        y=alt.Y("x:Q", title="xₜ")
    )
)

line = base.mark_line()

points = base.mark_point(
    filled=True,
    size=80
)

chart = (
    line + points
).properties(
    width=750,
    height=380,
    title="Linear Growth: Predictable, Stable, Smooth"
)

chart
```
