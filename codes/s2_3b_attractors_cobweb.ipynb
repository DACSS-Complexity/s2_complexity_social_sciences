```python
import numpy as np
import matplotlib.pyplot as plt

from matplotlib.animation import FuncAnimation
from IPython.display import HTML


# --------------------------------------------------
# 1. Logistic rule
# --------------------------------------------------

def logistic_curve(x, r):
    return r * x * (1 - x)


# --------------------------------------------------
# 2. Model settings
# --------------------------------------------------

r_value = 2.5
x0 = 0.1
n_iter = 30


# --------------------------------------------------
# 3. Create the cobweb data
# --------------------------------------------------

px = [x0]
py = [0]

x_t = x0

for t in range(n_iter):

    x_next = logistic_curve(x_t, r_value)

    # Vertical:
    # apply the rule x_(t+1) = f(x_t)
    px.append(x_t)
    py.append(x_next)

    # Horizontal:
    # x_(t+1) becomes the next x_t
    px.append(x_next)
    py.append(x_next)

    x_t = x_next


# --------------------------------------------------
# 4. Create the figure
# --------------------------------------------------

fig, ax = plt.subplots(figsize=(6, 6))

xs = np.linspace(0, 1, 500)

# Logistic curve
ax.plot(
    xs,
    logistic_curve(xs, r_value),
    linewidth=1.5,
    label=f"$f(x)={r_value}x(1-x)$"
)

# Identity line
ax.plot(
    xs,
    xs,
    '--',
    alpha=0.5,
    label="$x_{t+1}=x_t$"
)

# Empty cobweb: it will be filled step by step
cobweb_line, = ax.plot(
    [],
    [],
    linewidth=1.5
)

moving_point, = ax.plot(
    [],
    [],
    'o',
    markersize=5
)


ax.set_xlim(0, 1)
ax.set_ylim(0, 1)

ax.set_xlabel("$x_t$")
ax.set_ylabel("$x_{t+1}$")

ax.set_title(
    f"Cobweb Dynamics — r = {r_value}"
)

ax.legend(loc='upper left')


# --------------------------------------------------
# 5. Animation functions
# --------------------------------------------------

def init():

    cobweb_line.set_data([], [])
    moving_point.set_data([], [])

    return cobweb_line, moving_point


def update(frame):

    # Reveal the cobweb one step at a time
    cobweb_line.set_data(
        px[:frame + 1],
        py[:frame + 1]
    )

    # Current position
    moving_point.set_data(
        [px[frame]],
        [py[frame]]
    )

    return cobweb_line, moving_point


# --------------------------------------------------
# 6. Create animation
# --------------------------------------------------

ani = FuncAnimation(
    fig,
    update,
    frames=len(px),
    init_func=init,
    interval=500,
    blit=True,
    repeat=False
)

plt.close(fig)


# --------------------------------------------------
# 7. Display in Colab / Jupyter
# --------------------------------------------------

HTML(ani.to_jshtml())
```
