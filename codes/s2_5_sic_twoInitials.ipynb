```python
import numpy as np
import pandas as pd
import altair as alt
import ipywidgets as widgets

from IPython.display import display, clear_output


# -----------------------------------
# Controls
# -----------------------------------

r_slider = widgets.FloatSlider(
    value=3.7,
    min=2.5,
    max=3.99,
    step=0.01,
    description="r:",
    continuous_update=False,
    readout_format=".2f"
)

epsilon_slider = widgets.FloatLogSlider(
    value=0.001,
    base=10,
    min=-5,       # 10^-5
    max=-1,       # 10^-1
    step=0.1,
    description="ε:",
    continuous_update=False
)

plot_output = widgets.Output()


# -----------------------------------
# Draw the two representations
# -----------------------------------

def draw_parabolas(*args):

    r = r_slider.value
    epsilon = epsilon_slider.value

    x0_1 = 0.10
    x0_2 = x0_1 + epsilon

    x = np.linspace(0, 1, 500)
    y = r * x * (1 - x)

    label_1 = f"x₀ = {x0_1:.6f}"
    label_2 = f"x₀ = {x0_2:.6f}"

    # Same rule represented twice
    df_1 = pd.DataFrame({
        "x_t": x,
        "x_next": y,
        "x0": label_1
    })

    df_2 = pd.DataFrame({
        "x_t": x,
        "x_next": y,
        "x0": label_2
    })


    # Bottom parabola: thick solid
    bottom = (
        alt.Chart(df_1)
        .mark_line(
            strokeWidth=7
        )
        .encode(
            x=alt.X(
                "x_t:Q",
                title="xₜ"
            ),

            y=alt.Y(
                "x_next:Q",
                title="xₜ₊₁",
                scale=alt.Scale(domain=[0, 1])
            ),

            color=alt.Color(
                "x0:N",
                title="Initial condition",
                scale=alt.Scale(
                    domain=[label_1, label_2],
                    range=["steelblue", "orange"]
                )
            )
        )
    )


    # Top parabola: dotted
    top = (
        alt.Chart(df_2)
        .mark_line(
            strokeWidth=3,
            strokeDash=[2, 5]
        )
        .encode(
            x=alt.X(
                "x_t:Q",
                title="xₜ"
            ),

            y=alt.Y(
                "x_next:Q",
                title="xₜ₊₁",
                scale=alt.Scale(domain=[0, 1])
            ),

            color=alt.Color(
                "x0:N",
                title="Initial condition",
                scale=alt.Scale(
                    domain=[label_1, label_2],
                    range=["steelblue", "orange"]
                )
            )
        )
    )


    chart = (
        alt.layer(bottom, top)
        .properties(
            width=650,
            height=500,
            title=(
                f"Same rule — r = {r:.2f}, "
                f"ε = {epsilon:.6f}"
            )
        )
        .interactive()
    )


    with plot_output:
        clear_output(wait=True)
        display(chart)


# -----------------------------------
# Connect sliders
# -----------------------------------

r_slider.observe(
    draw_parabolas,
    names="value"
)

epsilon_slider.observe(
    draw_parabolas,
    names="value"
)


display(
    widgets.HBox([
        r_slider,
        epsilon_slider
    ]),
    plot_output
)

draw_parabolas()
```

```python
import numpy as np
import pandas as pd
import altair as alt
import ipywidgets as widgets

from IPython.display import display, clear_output


# -----------------------------------
# Logistic trajectory
# -----------------------------------

def logistic_trajectory(x0, r, n):
    values = [x0]

    for _ in range(n):
        values.append(
            r * values[-1] * (1 - values[-1])
        )

    return values


# -----------------------------------
# Controls
# -----------------------------------

r_slider = widgets.FloatSlider(
    value=3.7,
    min=2.5,
    max=3.99,
    step=0.01,
    description="r:",
    continuous_update=False,
    readout_format=".2f"
)

epsilon_slider = widgets.FloatLogSlider(
    value=0.001,
    base=10,
    min=-5,      # 10^-5
    max=-1,      # 10^-1
    step=0.1,
    description="ε:",
    continuous_update=False
)

plot_output = widgets.Output()


# -----------------------------------
# Draw chart
# -----------------------------------

def draw_trajectories(*args):

    r = r_slider.value
    epsilon = epsilon_slider.value

    x0_1 = 0.10
    x0_2 = x0_1 + epsilon

    n_iter = 50

    traj_1 = logistic_trajectory(
        x0_1, r, n_iter
    )

    traj_2 = logistic_trajectory(
        x0_2, r, n_iter
    )

    df = pd.DataFrame({
        "t": list(range(n_iter + 1)) * 2,

        "x": traj_1 + traj_2,

        "Initial condition": (
            [f"x₀ = {x0_1:.6f}"] * (n_iter + 1)
            +
            [f"x₀ = {x0_2:.6f}"] * (n_iter + 1)
        )
    })

    chart = (
        alt.Chart(df)
        .mark_line(
            point=True,
            strokeWidth=2
        )
        .encode(
            x=alt.X(
                "t:Q",
                title="Iteration"
            ),

            y=alt.Y(
                "x:Q",
                title="xₜ",
                scale=alt.Scale(
                    domain=[0, 1]
                )
            ),

            color=alt.Color(
                "Initial condition:N",
                title="Initial condition"
            ),

            strokeDash=alt.StrokeDash(
                "Initial condition:N",
                title="Initial condition"
            ),

            tooltip=[
                "Initial condition:N",
                alt.Tooltip(
                    "t:Q",
                    title="Iteration"
                ),
                alt.Tooltip(
                    "x:Q",
                    format=".8f"
                )
            ]
        )
        .properties(
            width=750,
            height=450,
            title=(
                f"Same rule — r = {r:.2f}, "
                f"ε = {epsilon:.6f}"
            )
        )
        .interactive()
    )

    with plot_output:
        clear_output(wait=True)
        display(chart)


# -----------------------------------
# Connect sliders
# -----------------------------------

r_slider.observe(
    draw_trajectories,
    names="value"
)

epsilon_slider.observe(
    draw_trajectories,
    names="value"
)


display(
    widgets.HBox([
        r_slider,
        epsilon_slider
    ]),
    plot_output
)

draw_trajectories()
```
